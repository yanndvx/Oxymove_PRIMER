## "State of the art" on the **efficiency** notion

- Taelman et al. (2011) : Ces travaux établissent une forte corrélation entre le déclin du contenu fréquentiel du signal sEMG (fréquence moyenne) et la pente de l'indice d'oxygénation tissulaire (TOI) mesuré par NIRS lors de l'apparition de la fatigue. Ils concluent que l'EMG et la NIRS mesurent deux phénomènes physiologiques complémentaires indispensables pour quantifier le rendement musculaire.
- Daniel et al. (2023 / 2026) : Cette recherche souligne que l'efficacité neuromusculaire ne peut plus être étudiée avec l'EMG isolé. L'approche multimodale (sEMG + mNIRS) est requise pour comprendre les mécanismes liant la production de force, en particulier excentrique, à la performance et à la fatigue.

## Ways to determine the pace
-> Using Minetti et al. (2002) equations
$$C_r(i) = 155{,}4 i^5 - 30{,}4 i^4 - 43{,}3 i^3 + 46{,}3 i^2 + 19{,}5 i + 3{,}6$$

*i* = inclinaison (à plat *i*=0)
C = Metabolic cost
r = running

à plat : 
$$C_r(0) = 3{,}60 \text{ J}\cdot\text{kg}^{-1}\cdot\text{m}^{-1}$$
+10% : $$C_r(+0{,}10) \approx 5{,}97 \text{ J}\cdot\text{kg}^{-1}\cdot\text{m}^{-1}$$
-10%: $$C_r(-0{,}10) \approx 2{,}15 \text{ J}\cdot\text{kg}^{-1}\cdot\text{m}^{-1}$$
La puissance métabolique cible $Pcible$= $Cr(0)*v(0)$ (v=vitesse en m/s) 
Pour un participant qui à RPE 25 à 10km/h (2.78m/s):
$Pcible = 3.6*2.78=10 W.Kg$

On converti pour les vitesse inclinées :
$$v_{i} = v_{\text{plat}} \times \left( \frac{Pcible}{C_r(i)} \right)$$
Soit : 
$$v_{+10\%}=\frac{Pcible}{Cr_{+10\%}}=\frac{10}{5.97}=1.68 m.s$$
1.68m/s x 3.6 = 6.0 km/h

$$v_{-10\%}=\frac{Pcible}{Cr_{-10\%}}=\frac{10}{2.15}=4.65 m.s$$
4.65m/s x 3.6 = 16.7 km/h


Pour un RPE 50 à 14km/h:
V+10% = 8.5 km/h
V-10% = 23.4 km/h
