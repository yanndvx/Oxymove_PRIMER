## 3. Cadre Théorique & Problématique

### 3.1. Objectifs du Projet
- Évaluer le rendement (*efficiency*) et le coût métabolique (fonction oxydative via mNIRS) lors de la course en régimes concentrique (+10%) et excentrique (-10%).
- Vérifier si une plus grande force excentrique des fléchisseurs du genou et une production de force rapide (*Rate of Force Development* - RFD) améliorent l'efficacité neuromusculaire en réduisant les demandes d'activation cumulées (sEMG).

### 3.2. Question Scientifique Principale
> Dans quelle mesure les tests de force excentrique et rapide prédisent-ils l'efficacité neuromusculaire et le coût métabolique (évalués par sEMG et mNIRS) lors de la course en régimes concentrique et excentrique ?

#### Sous-questions scientifiques :
1. La réduction des demandes d'activation cumulées, permise par une force excentrique supérieure, se vérifie-t-elle simultanément par une meilleure efficacité de la phosphorylation oxydative locale ?
2. L'avantage physiologique d'un profil de force élevé se maintient-il de manière identique lors d'une course en montée (+10%) et en descente (-10%) à différentes intensités métaboliques ?
3. L'ajout de contraintes externes, comme la variation de l'allure, modifie-t-il la relation entre la force des fléchisseurs et la dynamique d'oxygénation musculaire ?

### 3.3. État de l'Art sur la Notion de Rendement (*Efficiency*)

- **Taelman et al. (2011) :** Établissent une forte corrélation entre le déclin du contenu fréquentiel sEMG (fréquence moyenne) et la pente de l'indice d'oxygénation tissulaire (TOI) mesuré par NIRS lors de l'apparition de la fatigue. L'EMG et la NIRS mesurent deux phénomènes physiologiques complémentaires indispensables pour quantifier le rendement musculaire.
- **Daniel et al. (2023 / 2026) :** Soulignent que l'efficacité neuromusculaire ne peut plus être étudiée avec l'EMG isolé. L'approche multimodale (sEMG + mNIRS) est requise pour comprendre les mécanismes liant la production de force, en particulier excentrique, à la performance et à la fatigue.

*À définir précisément :* Définition opérationnelle complète de la *muscular efficiency*.

### 4.2. Choix de la Condition Expérimentale pour Manipuler la Force

#### Analyse des variables :
- **Pente (Vernillo et al., 2017) :** Les modifications de pente induisent des variations majeures des forces de réaction au sol, de l'activation neuromusculaire et du coût énergétique. *Limite :* La pente ne permet pas de faire varier la force au sein de la condition témoin à plat (0%).
- **Vitesse vs Gilet lesté :**

| Critères | Vitesse de course (Retenue) | Gilet lesté |
|---|---|---|
| **Justification scientifique** | Gold standard pour faire varier la charge métabolique et le RFD | Utilisé en biomécanique, mais peu adapté à l'efficience neuromusculaire |
| **Impact biomécanique** | Maintien d'une cinématique de course classique et naturelle | Risque d'altérer la posture du tronc/bassin et d'ajouter du bruit |
| **Faisabilité pratique** | Élevée et simple à mettre en œuvre sur tapis | Faisable mais nécessite attention avec l'équipement (Cosmed, etc.) |

#### Remarque technique Matériel :
- Les plateformes de force du tapis roulant ne sont pas exploitables lorsque le tapis est incliné (bruit provoqué par les vérins / calibration impossible en pente). Validation technique en cours (contact pris avec Simon).

### 4.3. Détermination des Allures de Course (Minetti et al., 2002)

Équation polynomiale du coût métabolique $C_r(i)$ en fonction de l'inclinaison $i$ :
$$C_r(i) = 155{,}4 i^5 - 30{,}4 i^4 - 43{,}3 i^3 + 46{,}3 i^2 + 19{,}5 i + 3{,}6$$

- **Plat ($i=0$) :** $C_r(0) = 3{,}60 \text{ J}\cdot\text{kg}^{-1}\cdot\text{m}^{-1}$
- **Montée (+10% / $i=+0{,}10$) :** $C_r(+0{,}10) \approx 5{,}97 \text{ J}\cdot\text{kg}^{-1}\cdot\text{m}^{-1}$
- **Descente (-10% / $i=-0{,}10$) :** $C_r(-0{,}10) \approx 2{,}15 \text{ J}\cdot\text{kg}^{-1}\cdot\text{m}^{-1}$

#### Calcul d'équivalence métabolique :
Puissance métabolique cible : $P_{\text{cible}} = C_r(0) \times v_{\text{plat}}$

Conversion des vitesses :
$$v_{i} = v_{\text{plat}} \times \left( \frac{C_r(0)}{C_r(i)} \right)$$

*Exemple pour $v_{\text{plat}} = 10\text{ km/h}$ ($2{,}78\text{ m/s}$) $\rightarrow P_{\text{cible}} = 10\text{ W/kg}$ :*
- $v_{+10\%} = \frac{10}{5{,}97} = 1{,}68\text{ m/s} \rightarrow \mathbf{6{,}0\text{ km/h}}$
- $v_{-10\%} = \frac{10}{2{,}15} = 4{,}65\text{ m/s} \rightarrow \mathbf{16{,}7\text{ km/h}}$

*Exemple pour $v_{\text{plat}} = 14\text{ km/h}$ :*
- $v_{+10\%} = 8{,}4\text{ km/h}$
- $v_{-10\%} = 23{,}4\text{ km/h}$

## 6. Contexte Scientifique (Dossier CERUM)

**Contexte général :** L’exercice excentrique est connu pour permettre un haut niveau de production de force, avec un rendement musculaire efficient. Bien que la force isocinétique des fléchisseurs du genou soit suggérés comme marqueur de l’efficacité neuromusculaire, on sait peu de choses sur leurs capacités prédictives pour des actions dynamiques concentriques et excentriques, comme celles impliquées lors de la course à pied. Aussi, une approche multimodale couplant électromyographie de surface (sEMG) et spectroscopie du proche infrarouge (mNIRS) pourrait s’avérer pertinente pour examiner la commande motrice et sa relation à la dynamique métabolique local de l’oxygénation.

**Contexte spécifique :** Des travaux récents suggèrent qu'un profil de force des fléchisseurs du genou en mode excentrique (évaluée via un test de contraction maximale volontaire), améliorerait l'efficience neuro-musculaire en réduisant les demandes d'activations cumulées (CNA) lors de la course à pied (Zhang et al., 2026). Toutefois, l'évaluation de cette notion d'efficience s'est historiquement limitée à l'analyse isolée de la commande motrice par électromyographie de surface (sEMG). 

Afin d'obtenir une vision globale des mécanismes sous-jasents, une approche multimodale est nécessaire (Daniel et al., 2026). Le couplage simultané du sEMG et de la spectroscopie du proche infrarouge appliqué au muscle (mNIRS) permet d'associer la commande motrice aux cinétiques d'oxygénation locales.

L’exercice excentrique est reconnu pour améliorer le rendement musculaire dans diverses activités dynamiques (Vogt et al., 2014). De plus il a été démontré que les profils de force sont corrélés à l’efficacité neuromusculaire, en réduisant les demandes d’activation cumulées (Zhang et al., 2026). Bien que la majorité des études évaluent les performances de course sur un tapis roulant plat, on sait peu de choses sur la capacité prédictive des profils de force en ce qui concerne la composante excentrique et concentrique de la course à pied. La relation entre le couple de force et l’efficacité neuromusculaire est généralement étudiée en utilisant l’électromyographie de manière isolée. Or, une étude récente (Daniel et al., 2026) a souligné la nécessité d’adopter une approche multimodale.
